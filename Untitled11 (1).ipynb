```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping

print("Loading MNIST dataset...")
mnist = tf.keras.datasets.mnist
(X_train, y_train), (X_test, y_test) = mnist.load_data()

X_train = X_train.reshape((-1, 28, 28, 1)).astype('float32') / 255.0
X_test = X_test.reshape((-1, 28, 28, 1)).astype('float32') / 255.0

print("Creating an improved CNN architecture...")
model = models.Sequential([
    layers.Conv2D(32, (3, 3), padding='same', activation='relu', input_shape=(28, 28, 1)),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.25),

    layers.Conv2D(64, (3, 3), padding='same', activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.25),

    layers.Conv2D(128, (3, 3), padding='same', activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.25),

    layers.Flatten(),
    layers.Dense(256, activation='relu'),
    layers.BatchNormalization(),
    layers.Dropout(0.5),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

print("Setting up the distortion generator (Augmentation)...")
datagen = ImageDataGenerator(
    rotation_range=15,
    width_shift_range=0.1,
    height_shift_range=0.1,
    zoom_range=0.15
)

early_stop = EarlyStopping(
    monitor='val_accuracy',
    patience=5,
    restore_best_weights=True
)

print("Training the neural network...")
model.fit(
    datagen.flow(X_train, y_train, batch_size=64),
    epochs=40,
    validation_data=(X_test, y_test),
    callbacks=[early_stop]
)

print("Model is ready! Run the evaluation on your photos!")
```

```python
import cv2
import os
import numpy as np
import matplotlib.pyplot as plt

def process_and_center_digit(img_path):
    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
    if img is None:
        return None

    margin_y = int(img.shape[0] * 0.05)
    margin_x = int(img.shape[1] * 0.05)
    img = img[margin_y:-margin_y, margin_x:-margin_x]

    img = cv2.resize(img, (500, int(500 * img.shape[0] / img.shape[1])))
    blurred = cv2.GaussianBlur(img, (11, 11), 0)
    thresh = cv2.adaptiveThreshold(blurred, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY_INV, 45, 6)

    kernel_dilate_balanced = np.ones((5, 5), np.uint8)
    thresh = cv2.dilate(thresh, kernel_dilate_balanced, iterations=1)

    contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    valid_contours = []
    h_img, w_img = thresh.shape
    for cnt in contours:
        x, y, w, h = cv2.boundingRect(cnt)

        if w < 10 or h < 10:
            continue
        if w > w_img * 0.95 or h > h_img * 0.95:
            continue

        valid_contours.append(cnt)

    if not valid_contours:
        return np.zeros((1, 28, 28, 1), dtype='float32')
    c = max(valid_contours, key=cv2.contourArea)

    x, y, w, h = cv2.boundingRect(c)
    pad = 20
    y1, y2 = max(0, y - pad), min(thresh.shape[0], y + h + pad)
    x1, x2 = max(0, x - pad), min(thresh.shape[1], x + w + pad)
    roi = thresh[y1:y2, x1:x2]

    target_size = 20
    h_roi, w_roi = roi.shape

    if max(h_roi, w_roi) == 0:
        return np.zeros((1, 28, 28, 1), dtype='float32')

    ratio = target_size / max(h_roi, w_roi)
    new_w = max(1, int(w_roi * ratio))
    new_h = max(1, int(h_roi * ratio))
    resized = cv2.resize(roi, (new_w, new_h), interpolation=cv2.INTER_AREA)

    final_canvas = np.zeros((28, 28), dtype=np.uint8)
    start_y = (28 - new_h) // 2
    start_x = (28 - new_w) // 2
    final_canvas[start_y:start_y+new_h, start_x:start_x+new_w] = resized
    kernel_final = np.ones((2, 2), np.uint8)
    final_canvas = cv2.dilate(final_canvas, kernel_final, iterations=1)

    return final_canvas.reshape(1, 28, 28, 1).astype('float32') / 255.0

data_dir = '/content/drive/MyDrive/sadas'
predictions_log = []

print("The neural network is analyzing the photos...")

for digit in range(10):
    folder_path = os.path.join(data_dir, str(digit))
    if not os.path.exists(folder_path):
        continue

    files = [f for f in os.listdir(folder_path) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

    for filename in files:
        full_path = os.path.join(folder_path, filename)
        processed_img = process_and_center_digit(full_path)

        if processed_img is not None:
            pred_raw = model.predict(processed_img, verbose=0)
            pred_digit = np.argmax(pred_raw)
            predictions_log.append({
                'real': digit,
                'predicted': pred_digit,
                'image': processed_img.reshape(28, 28)
            })
if predictions_log:
    correct = sum(1 for p in predictions_log if p['real'] == p['predicted'])
    total = len(predictions_log)
    accuracy = correct / total

    print("-" * 30)
    print(f"PHOTOS PROCESSED: {total}")
    print(f"FINAL ACCURACY: {accuracy:.2%}")
    print("-" * 30)

    plt.figure(figsize=(15, 6))
    for i, item in enumerate(predictions_log[:10]):
        plt.subplot(2, 5, i + 1)
        plt.imshow(item['image'], cmap='gray')
        color = 'green' if item['real'] == item['predicted'] else 'red'
        plt.title(f"Real: {item['real']}\nPred: {item['predicted']}", color=color)
        plt.axis('off')

    plt.tight_layout()
    plt.show()
```
